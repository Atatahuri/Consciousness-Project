```python
import numpy as np 
import matplotlib.pyplot as plt

x = np.linspace(-2,2,20)
y = np.linspace(-2,2,20)

X,Y = np.meshgrid(x,y)

print(X)
print(Y)

```

```python
U = X
V = -Y 

plt.quiver(X,Y,U,V)
plt.show()



```

```python
x = 1.0
y = 1.0
dt = 0.01
X = []
Y = []
for i in range(300): 
    X.append(x)
    Y.append(y)
    x = x + x*dt
    y = y - y*dt
    
plt.plot(X,Y, marker = ".")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("trajectories for Anosov Flow")
plt.show()

```
