```python
#Izhikevich Model 
#v is membrane potential variable
#u is recovery  variable
def ODE(a,b,c,d,I,u,v):
    vaxis = []
    uaxis = []
    dt = 0.05
    for i in range(1000):
        vaxis.append(v)
        uaxis.append(u)
        dv = 0.04*v**2 + 5*v + 140 - u + I
        du = a * (b*v - u)
        v = v + dv * dt
        u = u + du * dt 
        if v >= 30: 
            v = c
            u = u + d
    return uaxis, vaxis

uaxis,vaxis = ODE(0.02,0.2,-65,8,100,-13,-65)
time = np.arange(len(uaxis)) * dt
plt.plot(time, vaxis)
plt.xlabel("time")
plt.ylabel("Neuron Fires")
plt.title("Izhikevich Neuron Firing")
plt.show()
    
```
