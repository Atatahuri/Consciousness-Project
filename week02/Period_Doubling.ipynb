```python
def logistic(x,r):
    return r*x*(1- x)


def orbit(x,r): 
    orbit_list = []
    for i in range(1000): 
        orbit_list.append(x)
        x = logistic(x,r)
    return orbit_list[-100:]


tail = orbit(0.2,2.9)
print(tail)


```

```python
long_tail = []
def long_orbit(x,r):
    for i in range(10000):
        x = logistic(x,r)
        long_tail.append(x)
    return long_tail[-100:]

print(long_orbit(0.2, 3.445))
r = 3.44657

print(long_orbit(0.2,r)[0] - long_orbit(0.2,r)[2])
```

```python
r = 3.5

def derivative_logistic(x,r):
    return r*(1 - 2*x)

tail = long_orbit(0.2,r)
a = derivative_logistic(tail[0],r) * derivative_logistic(tail[1],r) * derivative_logistic(tail[2],r) * derivative_logistic(tail[3],r)

print(a)
```

```python
def cocycle(x,r,p):
    multiplier = 1
    tail = long_orbit(x,r)
    for i in range(p):
        multiplier = multiplier * derivative_logistic(tail[i],r)
    return multiplier

print(cocycle(0.2,3.5,4))
```

```python
import matplotlib.pyplot as plt
import numpy as np

def logistic_solution(x0, t, M):
    return (M*x0) / ((M-x0)*np.exp(-4*t) + x0)

taxis = np.linspace(-5, 5, 2000)
initial = np.linspace(-4, 10, 20)

for x0 in initial:
    y = logistic_solution(x0, taxis, 4)
    plt.plot(taxis, y)

plt.axhline(0, linestyle="--")
plt.axhline(4, linestyle="--")
plt.xlabel("t")
plt.ylabel("x(t)")
plt.ylim(-10, 10)
plt.show()
```

```python
for x0 in initial:
    y = logistic_solution(x0, taxis, 4)

    y[np.abs(y) > 20] = np.nan

    plt.plot(taxis, y)

plt.axhline(0, linestyle="--")
plt.axhline(4, linestyle="--")
plt.ylim(-10, 10)
plt.show()
```
