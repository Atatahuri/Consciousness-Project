```python
def logistic(x,r):
   return r*x*(1-x)

x = 0.1
r = 3.9

orbit = [x]

for i in range(50):
   x = logistic(x,r)
   orbit.append(x)

n = np.arange(len(orbit))
plt.plot(n,orbit)
plt.show()


plt.plot(n,orbit)
plt.xlabel("Iteration x")
plt.ylabel("x_n")
plt.title("Logistic map orbit for r = 3.9")

plt.show()

```

```python
#Bifurcation map
#parameter space creation

r = np.linspace(2.5,4.0,1000)
for i in range(1000):
        x = 0.1
        for k in range(1000):
            x = logistic(x,r[i])
            
        tail = []
        for j in range(100):
            x = logistic(x,r[i])
            tail.append(x)
        plt.scatter([r[i]] * len(tail), tail)


plt.show()
```

```python
r = np.linspace(2.5,4.0,1000)
for i in range(1000):
        x = 0.1
        for k in range(1000):
            x = logistic(x,r[i])
            
        tail = []
        x_plot = []
        r_plot = []
        for j in range(100):
            x = logistic(x,r[i])
            tail.append(x)
        r_plot.extend([r[i]] * len(tail))
        x_plot.extend(tail)
        plt.scatter(r_plot,x_plot)
plt.xlabel("r")
plt.ylabel("last 100 iterates of rx(1-x)")
plt.title("Bifurcation Diagram")
plt.show()





```
