```python
import numpy as np
import matplotlib.pyplot as plt


def logistic(x,r): 
     return r*x*(1-x)

def logistic_derivative(x,r):
     return r*(1 - 2*x)

print(logistic_derivative(0.6,2.5))
```

```python
def lyapunov(x, r, n):
    total = 0

    for j in range(1000):
        x = logistic(x,r)
        for i in range(n):
            total = total + np.log(np.abs(logistic_derivative(x, r)))
            x = logistic(x, r)
        return total / n

print(lyapunov(0.1, 2.5, 1000))
print(lyapunov(0.1, 3.2, 1000))
print(lyapunov(0.1, 3.9, 1000))
```

```python
import matplotlib.pyplot as plt

r_values = np.linspace(2.5,4.0,1000)


lambda_values = []

for r in r_values:
    lambda_values.append(lyapunov(0.1,r,1000))

plt.plot(r_values, lambda_values)
plt.plot(r_values,[0] * len(r_values))
plt.plot([3.84]*len(lambda_values), lambda_values)
plt.xlabel("r_values")
plt.ylabel("Lyapunov Exponents")
plt.title("Distribution of Lyapunov Exponents for logistic map with different constant values")

plt.show()

```

```python
x = 0.1
y = 0.100000001
r = 3.9
distances = []

for i in range(100): 
    x = logistic(x,r)
    y = logistic(y,r)
    distances.append(abs(x - y))

plt.semilogy(range(100),distances)
plt.xlabel("Iteration n")
plt.ylabel("|x_n - y_n|")
plt.title("Sensitivity to initial condition with parameter r")

plt.show()
```
